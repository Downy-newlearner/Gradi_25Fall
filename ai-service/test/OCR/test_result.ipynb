```python
def load_file_to_dict(file_path):
    """'{파일명}, {페이지번호}' 형식의 텍스트 파일을 dict로 변환"""
    data = {}
    with open(file_path, "r", encoding="utf-8") as f:
        for line in f:
            line = line.strip()
            if not line:
                continue
            if ',' in line:
                filename, value = line.split(',', 1)
                data[filename.strip()] = value.strip()
    return data


def compare_page_numbers(ground_truth_path, test_result_path):
    gt_dict = load_file_to_dict(ground_truth_path)
    test_dict = load_file_to_dict(test_result_path)

    differences = []

    for filename, gt_value in gt_dict.items():
        test_value = test_dict.get(filename)
        if test_value is None:
            differences.append((filename, gt_value, "파일 없음"))
        elif gt_value != test_value:
            differences.append((filename, gt_value, test_value))

    # test_result에만 존재하는 파일도 찾기
    for filename in test_dict:
        if filename not in gt_dict:
            differences.append((filename, "정답 없음", test_dict[filename]))

    return differences
```

```python
ground_truth_file = "images/page_number.txt"
test_result_file = "test_result/easyocr_page_number.txt"

diffs = compare_page_numbers(ground_truth_file, test_result_file)

if not diffs:
    print("✅ 모든 값이 일치합니다.")
else:
    print("❌ 서로 다른 값이 발견되었습니다:\n")
    for filename, gt, pred in diffs:
        print(f"{filename}\n  → 정답: {gt}\n  → 결과: {pred}\n")
```
